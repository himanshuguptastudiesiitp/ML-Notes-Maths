```python
import numpy as np
```

```python
# Generate synthetic Data  with noise y = 2.5*X +5
```

```python
np.random.seed(42)
```

```python
X= 2* np.random.rand(100,1) # 100 sample, 1 feature
Y= 2.5*X+5+np.random.randn(100,1) # b=5, w=2.5
```

```python
# initialize model parameters

w= 0.0
b=0.0
learning_rate=0.1
epochs=1000
N=len(X)
```

```python
# Training loop

for epoch in range(epochs):
  y_pred=w*X+b

  loss=np.mean((Y-y_pred)**2)

  #Calculate Gradients(partial Derivatives)

  dw=(-2/N)*np.sum(X*(Y-y_pred))
  db=(-2/N)*np.sum(Y-y_pred)

  #update Parameters
  w= w-learning_rate*dw
  b=b-learning_rate*db

  if (epoch +1)%200==0:
    print(f"Epoch {epoch+1}/{epoch} | Loss(MSE):{loss:.4f} | w:{w:.4f},b:{b:.4f}")



```

```python
print(f"\n Final Learned Formula: y= {w:.2f}*X+ {b:.2f}")
```

```python
# make prediction

X_new= np.array([[1.5],[3.0]])
predictions=w*X_new+b
for x_val,pred in zip(X_new.ravel(),predictions.ravel()):
  print(f"Input: {x_val:.2f} | Predicted Output: {pred:.2f}")
```
